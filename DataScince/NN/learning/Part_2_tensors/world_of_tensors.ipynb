Глубокое обучение состоит в создании системы, которая может преобразовывать данные из одного представления в другое. Это преобразование осуществляется путем нахождения закономерностей на множестве примеров, показывающих желаемую связь между входными и выходными данными. Например, система может отметить общую форму собаки и типичные цвета золотистого ретривера. Объединяя эти два свойства изображения, система может правильно сопоставлять изображения с данной формой и цветом с меткой золотистого ретривера, а не черного лабрадора (или, кстати, рыжего кота).

Получившаяся система может обрабатывать большие массивы схожих входных данных и генерировать значимый результат для этих данных. Процесс начинается с преобразования нашего входного сигнала в числа с плавающей запятой. В этой главе мы рассмотрим преобразование пикселей изображения в числа, но для этого мы сначала изучим, как работать со всеми числами с плавающей запятой в PyTorch с использованием тензоров. Тензоры являются фундаментальными структурами данных в фреймворках глубокого обучения, и PyTorch предоставляет богатый API для работы с ними, так что давайте начнем.
