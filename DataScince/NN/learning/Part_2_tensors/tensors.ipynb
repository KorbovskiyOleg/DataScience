Базовая арифметика со скалярами

```python
ones = torch.ones(2,2) + 2
ones
```

```python
twos = torch.ones(2,2)*3
twos
```

```python
threes = (torch.ones(2,2)*9-1)/2
threes
```

```python
fours = twos*2
fours
```

```python
sqrt2s = twos**0.5
sqrt2s
```
