```python
import numpy as np
```

```python
# функция активации
def act(x):
    return 0 if x < 0.5 else 1
```

```python
# функция , которая принимает входной сигнал пропускае через нейронную сеть, агрегирует его и выводит образовалась симпатия или нет(0 или 1)
def go(house,rock,attr):
    x = np.array([house,rock,attr]) # форммируем вектор входного сигнала
    w11 = [0.3,0.3,0]
    w12 = [0.4,-0.5,1]
    weight1 = np.array([w11,w12]) # матрица весов скрытого слоя
    weight2 = np.array([-1,1]) # вектор весов выходного слоя 

    sum_hidden = np.array(weight1, x) # вычисление суммы на входах нейронов скрытого слоя
    print("Значение сумм нейронов скрытого слоя : "+ str(sum_hidden))

    out_hidden = np.array([act(x) for x in sum_hidden])
    print("Значения на выходах нейронов скрытого слоя : "+ str(out_hidden))

    sum_end = np.dot(weight2, out_hidden)
    y = act(sum_end)
    print("Выходное значение : "+ str(y))
    


    

    print("Выходное значение : "+ str(y))
    
    return y
    
```

```python
# подаем на вход сигналы

house = 1
rock  = 0
attr = 1
```

```python
res = go(house,rock, attr)
if res == 1:
    print("Ты мне нравишся")
else:
    print("Созвонимся)")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# Исходные данные
x = np.array([1, 0, 1])
w11 = [0.3, 0.3, 0]
w12 = [0.4, -0.5, 1]
weight1 = np.array([w11, w12])

print("="*60)
print("Исходные данные:")
print("="*60)
print(f"Входной вектор x: {x}")
print(f"               = [x₁, x₂, x₃]")
print()
print(f"Матрица весов weight1:")
print(weight1)
print()
print(f"  Строка 1 (нейрон 1): {w11} = [w₁₁, w₁₂, w₁₃]")
print(f"  Строка 2 (нейрон 2): {w12} = [w₂₁, w₂₂, w₂₃]")
print("="*60)

# Вычисление
sum_hidden = np.dot(weight1, x)

print("\nВычисление через матричное умножение:")
print("="*60)

# Подробное вычисление для каждого нейрона
print(f"\nНейрон 1 (первая строка матрицы):")
print(f"  sum₁ = w₁₁·x₁ + w₁₂·x₂ + w₁₃·x₃")
print(f"  sum₁ = {w11[0]:.1f}·{x[0]} + {w11[1]:.1f}·{x[1]} + {w11[2]:.1f}·{x[2]}")
print(f"  sum₁ = {w11[0]*x[0]:.1f} + {w11[1]*x[1]:.1f} + {w11[2]*x[2]:.1f}")
print(f"  sum₁ = {sum_hidden[0]:.1f}")

print(f"\nНейрон 2 (вторая строка матрицы):")
print(f"  sum₂ = w₂₁·x₁ + w₂₂·x₂ + w₂₃·x₃")
print(f"  sum₂ = {w12[0]:.1f}·{x[1]} + {w12[1]:.1f}·{x[1]} + {w12[2]:.1f}·{x[2]}")
print(f"  sum₂ = {w12[0]*x[0]:.1f} + {w12[1]*x[1]:.1f} + {w12[2]*x[2]:.1f}")
print(f"  sum₂ = {sum_hidden[1]:.1f}")

print("\n" + "="*60)
print(f"Результат: sum_hidden = {sum_hidden}")
print("="*60)

# Создаём визуализацию
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# ===== 1. Матрица весов =====
ax1 = axes[0, 0]
im1 = ax1.imshow(weight1, cmap='Blues', aspect='auto', vmin=-1, vmax=1)
ax1.set_title('Матрица весов (2×3)\n[нейрон, вход]', fontsize=13, fontweight='bold', pad=20)

# Подписи ячеек
for i in range(weight1.shape[0]):
    for j in range(weight1.shape[1]):
        ax1.text(j, i, f'{weight1[i, j]:.1f}', 
                ha='center', va='center', fontsize=14, fontweight='bold',
                color='black' if abs(weight1[i, j]) < 0.5 else 'white')

ax1.set_xticks([0, 1, 2])
ax1.set_xticklabels([f'x₁\n(вход 1)', f'x₂\n(вход 2)', f'x₃\n(вход 3)'], fontsize=11)
ax1.set_yticks([0, 1])
ax1.set_yticklabels([f'Нейрон 1\n(w₁₁, w₁₂, w₁₃)', f'Нейрон 2\n(w₂₁, w₂₂, w₂₃)'], fontsize=11)

# Добавляем цветовую шкалу
cbar1 = plt.colorbar(im1, ax=ax1, fraction=0.046, pad=0.04)
cbar1.set_label('Значение веса', fontsize=10)

# ===== 2. Вектор входов =====
ax2 = axes[0, 1]
im2 = ax2.imshow(x.reshape(-1, 1), cmap='Greens', aspect='auto', vmin=0, vmax=1)
ax2.set_title('Вектор входов (3×1)\n[вход]', fontsize=13, fontweight='bold', pad=20)

for i in range(len(x)):
    ax2.text(0, i, f'{x[i]}', 
            ha='center', va='center', fontsize=16, fontweight='bold')

ax2.set_xticks([0])
ax2.set_xticklabels(['Значение'], fontsize=11)
ax2.set_yticks([0, 1, 2])
ax2.set_yticklabels([f'x₁ = {x[0]}', f'x₂ = {x[1]}', f'x₃ = {x[2]}'], fontsize=11)

cbar2 = plt.colorbar(im2, ax=ax2, fraction=0.046, pad=0.04)
cbar2.set_label('Значение входа', fontsize=10)

# ===== 3. Результат (вектор сумм) =====
ax3 = axes[1, 0]
im3 = ax3.imshow(sum_hidden.reshape(-1, 1), cmap='Oranges', aspect='auto', vmin=-1, vmax=1)
ax3.set_title('Результат: суммы на нейронах (2×1)\n[нейрон]', fontsize=13, fontweight='bold', pad=20)

for i in range(len(sum_hidden)):
    ax3.text(0, i, f'{sum_hidden[i]:.1f}', 
            ha='center', va='center', fontsize=16, fontweight='bold',
            color='black' if abs(sum_hidden[i]) < 0.5 else 'white')

ax3.set_xticks([0])
ax3.set_xticklabels(['Значение'], fontsize=11)
ax3.set_yticks([0, 1])
ax3.set_yticklabels([f'Нейрон 1:\n{sum_hidden[0]:.1f}', f'Нейрон 2:\n{sum_hidden[1]:.1f}'], fontsize=11)

cbar3 = plt.colorbar(im3, ax=ax3, fraction=0.046, pad=0.04)
cbar3.set_label('Сумма (без активации)', fontsize=10)

# ===== 4. Схема умножения =====
ax4 = axes[1, 1]
ax4.axis('off')

# Рисуем схему
y_pos = 0.9
ax4.text(0.05, y_pos, 'Матричное умножение:', fontsize=14, fontweight='bold', transform=ax4.transAxes)

y_pos -= 0.1
ax4.text(0.05, y_pos, '┌─┐   ┌─┐', fontsize=12, fontweight='bold', transform=ax4.transAxes)
y_pos -= 0.05
ax4.text(0.05, y_pos, '│w│ × │x│ = │sum│', fontsize=14, fontweight='bold', transform=ax4.transAxes)
y_pos -= 0.05
ax4.text(0.05, y_pos, '└─┘   └─┘', fontsize=12, fontweight='bold', transform=ax4.transAxes)

y_pos -= 0.15
ax4.text(0.05, y_pos, 'Детально:', fontsize=13, fontweight='bold', transform=ax4.transAxes)

y_pos -= 0.1
ax4.text(0.05, y_pos, f'Нейрон 1: {w11[0]:.1f}×{x[0]} + {w11[1]:.1f}×{x[1]} + {w11[2]:.1f}×{x[2]} = {sum_hidden[0]:.1f}', 
         fontsize=12, transform=ax4.transAxes, color='blue')

y_pos -= 0.08
ax4.text(0.05, y_pos, f'Нейрон 2: {w12[0]:.1f}×{x[0]} + {w12[1]:.1f}×{x[1]} + {w12[2]:.1f}×{x[2]} = {sum_hidden[1]:.1f}', 
         fontsize=12, transform=ax4.transAxes, color='red')

y_pos -= 0.15
ax4.text(0.05, y_pos, 'Правило:', fontsize=13, fontweight='bold', transform=ax4.transAxes)
y_pos -= 0.08
ax4.text(0.05, y_pos, 'Каждая СТРОКА матрицы весов', fontsize=11, transform=ax4.transAxes)
y_pos -= 0.06
ax4.text(0.05, y_pos, 'умножается на вектор входов →', fontsize=11, transform=ax4.transAxes)
y_pos -= 0.06
ax4.text(0.05, y_pos, 'скалярное произведение (dot product)', fontsize=11, transform=ax4.transAxes)

# Добавляем рамки для выделения
for ax in axes.flat:
    for spine in ax.spines.values():
        spine.set_linewidth(2)

plt.tight_layout()
plt.savefig('matrix_multiplication_visualization.png', dpi=150, bbox_inches='tight')
plt.show()

# Дополнительная наглядная схема в текстовом виде
print("\n" + "="*70)
print("ТЕКСТОВАЯ СХЕМА УМНОЖЕНИЯ")
print("="*70)
print()
print("  ┌─────────────────────────────────────┐")
print("  │  МАТРИЦА ВЕСОВ (2×3)                │")
print("  │  ┌───────────────────────────────┐  │")
print(f"  │  │  {w11[0]:5.1f}  {w11[1]:5.1f}  {w11[2]:5.1f}  │ ← Строка 1 (нейрон 1)")
print(f"  │  │  {w12[0]:5.1f}  {w12[1]:5.1f}  {w12[2]:5.1f}  │ ← Строка 2 (нейрон 2)")
print("  │  └───────────────────────────────┘  │")
print("  └─────────────────────────────────────┘")
print("              ↓")
print("              × (матричное умножение)")
print("              ↓")
print("  ┌─────────────────────────────────────┐")
print("  │  ВЕКТОР ВХОДОВ (3×1)                │")
print("  │  ┌───────┐                          │")
print(f"  │  │  {x[0]:5.1f}  │ ← x₁ (вход 1)           │")
print(f"  │  │  {x[1]:5.1f}  │ ← x₂ (вход 2)           │")
print(f"  │  │  {x[2]:5.1f}  │ ← x₃ (вход 3)           │")
print("  │  └───────┘                          │")
print("  └─────────────────────────────────────┘")
print("              ↓")
print("              =")
print("              ↓")
print("  ┌─────────────────────────────────────┐")
print("  │  РЕЗУЛЬТАТ (2×1)                    │")
print("  │  ┌───────┐                          │")
print(f"  │  │  {sum_hidden[0]:5.1f}  │ ← Сумма нейрона 1   │")
print(f"  │  │  {sum_hidden[1]:5.1f}  │ ← Сумма нейрона 2   │")
print("  │  └───────┘                          │")
print("  └─────────────────────────────────────┘")
print()
print("="*70)
print("ПОДРОБНОЕ ВЫЧИСЛЕНИЕ")
print("="*70)
print()
print("Нейрон 1:")
print("  " + "─" * 50)
print(f"  sum₁ = w₁₁·x₁ + w₁₂·x₂ + w₁₃·x₃")
print(f"       = {w11[0]:.1f}·{x[0]} + {w11[1]:.1f}·{x[1]} + {w11[2]:.1f}·{x[2]}")
print(f"       = {w11[0]*x[0]:.1f} + {w11[1]*x[1]:.1f} + {w11[2]*x[2]:.1f}")
print(f"       = {sum_hidden[0]:.1f}")
print("  " + "─" * 50)
print()
print("Нейрон 2:")
print("  " + "─" * 50)
print(f"  sum₂ = w₂₁·x₁ + w₂₂·x₂ + w₂₃·x₃")
print(f"       = {w12[0]:.1f}·{x[0]} + {w12[1]:.1f}·{x[1]} + {w12[2]:.1f}·{x[2]}")
print(f"       = {w12[0]*x[0]:.1f} + {w12[1]*x[1]:.1f} + {w12[2]*x[2]:.1f}")
print(f"       = {sum_hidden[1]:.1f}")
print("  " + "─" * 50)
print()
print("="*70)
print("РЕЗУЛЬТАТ:")
print("="*70)
print(f"  sum_hidden = [{sum_hidden[0]:.1f}, {sum_hidden[1]:.1f}]")
print("="*70)
```
