Ок! Видеокарта подключена и работает! Теперь сравним разницу в вычислениях!

```python
big_tensor = torch.rand(10000,10000) # создадим матрицу 10000 на 10000 
```

```python
# вычисления на CPU 
start = time.time()
cpu_res = big_tensor @ big_tensor
print(f"CPU время : {time.time()-start:.4f} сек")
```

```python
# вычисления на GPU

gpu_tensor = big_tensor.cuda()
start = time.time()
gpu_res = gpu_tensor @ gpu_tensor
torch.cuda.synchronize()
print(f"GPU время : {time.time()-start:.4f} сек")

```

```python
del gpu_res, gpu_tensor
torch.cuda.empty_cache()
```
