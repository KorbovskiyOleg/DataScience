```python
# импотр библиотек
from llama_cpp import Llama
import os
import sys
```

```python
# загрузка модели  Llama-2-7B-Chat_Q8_0
model_path = r"C:\Users\Админ\.ai-navigator\models\meta-llama\llama-2-7b-chat-hf\Llama-2-7B-Chat_Q8_0.gguf"
```

```python
# иницилизация модели
if os.path.exists(model_path):
    print("🎮 Загружаю модель на RTX 3050...")
    
    llm = Llama(
        model_path=model_path,
        n_ctx=2048,
        n_gpu_layers=25,      # 👈 Явно указываем 20 слоя на GPU
        n_threads=12,         # колличество потоков , зависит от проца
        #main_gpu=0,           # 👈 Используем основной GPU
        #tensor_split=[0.9],   # 👈 90% нагрузки на GPU
        low_vram=False,       # 👈 включить/отключить экономить видеопамять
        verbose=True         # 👈 Следим за логами

    )
    print("Модель загужена!Работаем!!!")
else:
    print("❌ Файл модели не найден")
```

```python

# функция для формирования запросов и ответов
def chat_with_llama(system_prompt, user_message, temperature=0.7):
    # Форматируем промпт в правильном формате
    prompt = f"""[INST] <<SYS>>
{system_prompt}
<</SYS>>

{user_message} [/INST]"""
    
    # Генерируем ответ
    response = llm(
        prompt,
        max_tokens=400,
        temperature=temperature,
        top_p=0.9,
        stop=["</s>", "[INST]"],
        echo=False,
        stream=False
    )
    
    return response['choices'][0]['text'].strip()
```

```python
# системный промпт 
system_prompt = """You're a useful AI assistant. Answer accurately and politely.Always answer in Russian. """
```

```python
# запрос 
user_message = "Объясни, что такое искусственный интеллект, простыми словами."

# ответ модели
answer = chat_with_llama(system_prompt, user_message)
print("Ответ модели:", answer)
```
