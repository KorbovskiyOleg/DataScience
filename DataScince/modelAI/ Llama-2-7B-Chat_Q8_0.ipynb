```python
from llama_cpp import Llama
import os
```

```python
# модель DeepSeek-Coder-V2-Lite-Instruct_Q4_K_M - тупая!!!
#model_path = r"C:\Users\Админ\.ai-navigator\models\deepseek-ai\DeepSeek-Coder-V2-Lite-Instruct\DeepSeek-Coder-V2-Lite-Instruct_Q4_K_M.gguf"

# модель DeepSeek-Coder-7B-Instruct-v1.5_Q4_K_M - тупая!!! 
#model_path = r"C:\Users\Админ\.ai-navigator\models\deepseek-ai\deepseek-coder-7b-instruct-v1.5\DeepSeek-Coder-7B-Instruct-v1.5_Q4_K_M.gguf"

# модель Llama-2-7B-Chat_Q8_0 - пока что лучшая!!!
model_path = r"C:\Users\Админ\.ai-navigator\models\meta-llama\llama-2-7b-chat-hf\Llama-2-7B-Chat_Q8_0.gguf"
```

```python
if os.path.exists(model_path):
    print("🎮 Загружаю модель на RTX 3050...")
    
    llm = Llama(
        model_path=model_path,
        n_ctx=2048,
        n_gpu_layers=25,      # 👈 Явно указываем 20 слоя на GPU
        n_threads=12,         # колличество потоков , зависит от проца
        #main_gpu=0,           # 👈 Используем основной GPU
        #tensor_split=[0.9],   # 👈 90% нагрузки на GPU
        low_vram=False,       # 👈 включить/отключить экономить видеопамять
        verbose=True         # 👈 Следим за логами

    )
    print("Модель загужена!Работаем!!!")
else:
    print("❌ Файл модели не найден")
```

```python
# Тестируем
response = llm(
    """You're a useful AI assistant. Answer accurately and politely.

        User: Сколько всего океанов на земле:

        Assistant:""",
        max_tokens=350,
        temperature=0.6,      # Случайность (0-1)
        top_p=0.3,           # Nucleus sampling
        repeat_penalty=1.1   # Штраф за повторения
        
    )
    
print("📝 Результат:")
print(response['choices'][0]['text'])
```
